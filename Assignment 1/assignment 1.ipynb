```python
#Problem 1

text = input("Enter a string: ")

letters = []
numbers = []

for char in text:
    if char.isdigit():
        numbers.append(int(char))
    elif char.isalpha():
        letters.append(char)

total = sum(numbers)

print("Word:", "".join(letters))
print("Sum:", total)
```

```python
#Problem 2

password = input("Enter a password: ")
special_characters = ["!", "@", "#", "$"]
if  len(password) < 8 or len(password) > 16:
    print("Password must be within 8 to 16 characters.")
elif not any(char.isupper() for char in password):
    print("Password must contain at least one uppercase.")
elif not any(char in special_characters for char in password):
    print("Password must contain at least one special character.")
else:
    print("password is secure.")
```

```python
# Problem 7
number = int(input("Enter a multi-digit number: "))

while number >= 10:
    total = 0

    while number > 0:
        digit = number % 10
        total += digit
        number //= 10

    number = total

print("Digital Root:", number)
```

```python
#problem 8

numbers = [1, 1, 2, 3, 3, 3, 4]
result = []
for number in numbers:
    if number not in [item[0] for item in result]:
        frequency = numbers.count(number)
        result.append((number, frequency))
print("Result:", result)
```

```python
# Problem 9

number = int(input("Enter a number: "))

factorial = 1
for i in range(1, number + 1):
    factorial *= i

total = 0
for i in range(1, number + 1):
    total += i

if factorial % total == 0:
    print(f"Factorial is {factorial}, Sum is {total}. {factorial} is divisible by {total}.")
else:
    print(f"Factorial is {factorial}, Sum is {total}. {factorial} is not divisible by {total}.")
```

```python
#problem 10
items_prices = {
    "apple" : 0.5,
    "bread" : 2.0,
    "milk" : 1.5
}

cart = []

while True:
    action = input("Enter `add` or `remove` or `done` ")
    if action == "done":
        break
    elif action == "add":
        item = input("Enter the item you want to add: ")
        if item in items_prices:
            cart.append(item)
        else:
            print("Item not found.")
    elif action == "remove":
        item = input("Enter the item you want to remove: ")
        if item in cart:
            cart.remove(item)
        else:
            print("Item not in cart.")
    else:
        print("Invalid action.")
print ("total bill: ", sum(items_prices[item] for item in cart))
    
```

```python
#problem 13
list = []
quantity = int(input("Enter the number of elements: "))
for i in range(quantity):
    element = int(input(f"Enter element {i + 1}: "))
    list.append(element)

peaks = []

for i in range(1, len(list) - 1):
    if list[i] > list[i - 1] and list[i] > list[i + 1]:
        peaks.append(list[i])

print("Peaks:", peaks)
```

```python
#Problem 14
original = {'A': 1, 'B': 2, 'C': 1}
flipped = {}

for key, value in original.items():
    if value in flipped:
        flipped[value].append(key)
    else:
        flipped[value] = [key]

print(flipped)
```

```python
#problem 15
string = input("Enter a string: ")
vowels = "aeiouAEIOU"
result = {}
for i in range(len(string)):
    char = string[i]
    if char in vowels:
      if char in result:
        result[char].append(i)
      else:
        key = char
        values = [i]
        result[key] = values

print(result)
```
