```python
#problem 1

def filter_strings(words, K):
    result = []

    for word in words:
        if len(word) > K:
            result.append(word)

    return result


words = ["apple", "pie", "banana"]
K = 3

print(filter_strings(words, K))
```

```python
#problem 2

def area(shape, *dimensions):
    if shape == "circle":
        radius = dimensions[0]
        result = 3.14 * radius ** 2
        return f"Circle Area: {result}"

    elif shape == "rectangle":
        length = dimensions[0]
        width = dimensions[1]
        result = length * width
        return f"Rectangle Area: {result}"

    else:
        return "Invalid shape"


print(area("circle", 5))
print(area("rectangle", 4, 6))
```

```python
#problem 5


class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, customer, table_number):
        reservation = {
            "customer": customer,
            "table": table_number
        }
        self.book_table.append(reservation)

    def customer_order(self, customer, item):
        order = {
            "customer": customer,
            "item": item
        }
        self.customer_orders.append(order)

    def print_menu(self):
        print("Menu:")
        print(self.menu_items)

    def print_reservations(self):
        print("Table Reservations:")
        print(self.book_table)

    def print_orders(self):
        print("Customer Orders:")
        print(self.customer_orders)



restaurant = Restaurant()


restaurant.add_item_to_menu("Burger", 150)
restaurant.add_item_to_menu("Pizza", 200)
restaurant.add_item_to_menu("Pasta", 120)


restaurant.book_tables("Ahmed", 1)
restaurant.book_tables("Mona", 2)


restaurant.customer_order("Ahmed", "Burger")
restaurant.customer_order("Mona", "Pizza")


restaurant.print_menu()
restaurant.print_reservations()
restaurant.print_orders()
```

```python
#problem 6


class Customer:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    def get_name(self):
        return self._name

    def set_name(self, name):
        self._name = name

    def get_age(self):
        return self._age

    def set_age(self, age):
        self._age = age


customer1 = Customer("Ahmed", 20)

print(customer1.get_name())
print(customer1.get_age())

customer1.set_age(21)
print(customer1.get_age())
```
